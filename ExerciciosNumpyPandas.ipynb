**Exercício 09 — Concatenando períodos**

Uma empresa armazena suas vendas mensalmente.

jan = pd.DataFrame({
    "produto": ["Notebook", "Mouse", "Teclado"],
    "valor": [3500, 100, 150]
})

fev = pd.DataFrame({
    "produto": ["Notebook", "Mouse", "Monitor"],
    "valor": [3200, 120, 1100]
})

Tarefa:
1. Utilize pd.concat() para criar uma única tabela com janeiro e fevereiro.
2. Faça o índice ser reiniciado após a concatenação.
3. Calcule o valor total vendido.
4. Calcule o valor total vendido por produto.
5. Identifique o produto com maior faturamento acumulado.


**9.1 - Utilizando pd.concat() para criar uma única tabela com janeiro e fevereiro.**

**Criando DF**

```python
jan = pd.DataFrame({ "produto": ["Notebook", "Mouse", "Teclado"], "valor": [3500, 100, 150] })
```

```python
fev = pd.DataFrame({ "produto": ["Notebook", "Mouse", "Monitor"], "valor": [3200, 120, 1100] })
```

**Criando tabela única com janeiro e fevereiro**

```python
vendas = pd.concat([jan, fev])

print(vendas)
```

**9.2 - Reniciando o índice após a concatenação.**

```python
vendas = pd.concat([jan, fev], ignore_index=True)

print(vendas)
```

**Note os números a esquerda (índice) 0, 1, 2, 3, 4, 5 foram organizados**

**9.3 - Calculando o valor total vendido.**

**9.5 - Identificando o produto com maior faturamento acumulado**

```python
maior_faturamento = total_por_produto.idxmax()

print(maior_faturamento)
```

pd.concat() → junta DataFrames

ignore_index=True → refaz o índice

.groupby() → separa por grupo

.idxmax() → encontra quem possui o maior valor

**Exercício 10 — Tabela dinâmica de vendas**

**Criando DF**

```python
df = pd.DataFrame({ "uf": ["SC", "SC", "PR", "PR", "SC", "RS", "RS", "PR"], "mes": ["Jan", "Fev", "Jan", "Fev", "Mar", "Jan", "Fev", "Mar"],
                   "valor": [500, 700, 300, 450, 800, 200, 350, 600] })
```

**Criando tabela dinâmica**

df = pd.DataFrame({
    "uf": ["SC", "SC", "PR", "PR", "SC", "RS", "RS", "PR"],
    "mes": ["Jan", "Fev", "Jan", "Fev", "Mar", "Jan", "Fev", "Mar"],
    "valor": [500, 700, 300, 450, 800, 200, 350, 600]
})

Utilize pivot_table() para criar uma tabela na qual:

• cada linha represente uma UF;
• cada coluna represente um mês;
• os valores sejam a soma das vendas;
• meses sem vendas apareçam com 0.

Depois:

1. Identifique a UF com maior venda total.
2. Identifique o mês com maior venda total.
3. Explique, em uma frase, qual é a vantagem dessa visualização para análise dos dados.


**10.1 - Identificando a UF com maior venda total**

```python
tabela = df.pivot_table(
    index="uf",
    columns="mes",
    values="valor",
    aggfunc="sum",
    fill_value=0
)

print(tabela)
```

**Explicando o código:**

Pense no **pivot_table()** como alguém organizando uma bagunça:

**index="uf"** → coloca as UFs nas linhas

**columns="mes"** → coloca os meses nas colunas

**values="valor"** → mostra os valores das vendas

**aggfunc="sum"** → se houver mais de uma venda, soma

**fill_value=0** → onde não houver venda, coloca 0

**Identificando a UF com maior venda total**

```python
total_uf = tabela.sum(axis=1)

print(total_uf)
```

**10.2 - Identifique o mês com maior venda total.**

```python
total_mes = tabela.sum(axis=0)

print(total_mes)
```

**10.3 - Explique, em uma frase, qual é a vantagem dessa visualização para análise dos dados.**

A tabela dinâmica facilita a comparação das vendas entre diferentes UFs e meses, permitindo identificar rapidamente onde e quando as vendas foram maiores.
